## 클래스

```python
class Account:
    balance = 100 # 클래스 변수. 모든 객체가 같은 변수를 가리킴
    # def __init__(self): 
    #     self.balance = 100 멤버 변수(속성)
    def deposit(self, amount):
        self.balance += amount
    def withdraw(self, amount):
        self.balance -= amount

a1 = Account()
a1.deposit(100)
a1.withdraw(50)
a2 = Account()

print(a1.balance)
print(Account.balance)
print(a2.balance)

print(Account.balance.__hash__)
print(a1.balance.__hash__) # 한번 a1.balance로 사용하면 따로 할당됨
print(a2.balance.__hash__)
```
