```python
import pandas as pd
import ast
import re
import difflib

# 유사도 연산 속도 향상을 위한 캐시 딕셔너리
fuzzy_cache = {}

# =====================================================================
# 2. 통합, 중복 제거 및 유사도 보정 함수
# =====================================================================
def merge_dedup_and_fuzzy_skills(row):
    def extract_list(val):
        if isinstance(val, list): return val
        if pd.isna(val): return []
        if isinstance(val, str):
            val = val.strip()
            if not val: return []
            if val.startswith('['):
                try: return ast.literal_eval(val)
                except: pass
            return [item.strip() for item in val.replace('(', ',').replace(')', ',').replace('/', ',').split(',')]
        return []

    combined_skills = extract_list(row.get('skill_tags', [])) + extract_list(row.get('hard_skills', []))
    final_skills = set()
    
    for skill in combined_skills:
        if not skill or pd.isna(skill): continue
            
        t_lower = str(skill).lower()
        t_normalized = re.sub(r'[^a-z0-9+#\.]', '', t_lower).strip()
        
        if not t_normalized: continue
            
        # [Step 1] 화이트리스트에 완벽하게 일치하는 경우 (가장 빠름)
        if t_normalized in MASTER_TECH_DICT:
            final_skills.add(MASTER_TECH_DICT[t_normalized])
            continue
            
        # [Step 2] 완벽히 일치하지 않지만, 단어 길이가 3 이상인 경우에 한해 80% 유사도 검사
        if len(t_normalized) >= 3:
            # 캐시 확인 (이미 검사해본 단어인지)
            if t_normalized in fuzzy_cache:
                if fuzzy_cache[t_normalized]:
                    final_skills.add(MASTER_TECH_DICT[fuzzy_cache[t_normalized]])
                continue
            
            # 화이트리스트의 'Key'들과 80%(cutoff=0.8) 이상 비슷한지 검사
            matches = difflib.get_close_matches(t_normalized, MASTER_TECH_DICT.keys(), n=1, cutoff=0.8)
            
            if matches:
                best_match = matches[0]
                fuzzy_cache[t_normalized] = best_match # 성공 캐싱
                final_skills.add(MASTER_TECH_DICT[best_match])
                # print(f"💡 오타 보정 성공: {skill} -> {MASTER_TECH_DICT[best_match]}") # 필요시 주석 해제하여 확인
            else:
                fuzzy_cache[t_normalized] = None # 실패 캐싱 (다음번엔 검사 안하고 바로 버림)
                
        # 1~2글자인데 매핑 사전에 없다? (예: 'A', '10') -> 조용히 버려짐
        # 길이가 3 이상인데 80% 유사도에도 안 걸린다? (예: 'excel', '100g') -> 조용히 버려짐

    return list(final_skills)


# =====================================================================
# 3. 데이터프레임 적용 및 테스트
# =====================================================================
# 오타와 쓰레기가 섞인 악랄한 테스트 데이터
test_data = {
    'location_perfect': ['서울 강남구'],
    'skill_tags': ["['Git', 'reaact', 'Node.js', 'Typesccript', '100G']"], # react, typescript 오타 + 100G(쓰레기)
    'hard_skills': [['Next.js', 'React', 'postgressql', 'Firebase', 'Excel']] # postgresql 오타 + Excel(쓰레기)
}
df_test = pd.DataFrame(test_data)

print("⏳ 기술 스택 통합, 오타 교정 및 중복 제거를 시작합니다...")
df_test['final_tech_stack'] = df_test.apply(merge_dedup_and_fuzzy_skills, axis=1)

pd.set_option('display.max_colwidth', None)
display(df_test[['location_perfect', 'skill_tags', 'hard_skills', 'final_tech_stack']])
```

```python
import pandas as pd
import ast
import re
import json
import difflib  # 파이썬 내장 유사도 매칭 라이브러리

# 1. 원티드 표준 단어장 로드
def load_wanted_vocab(filepath='wanted_techs.json'):
    standard_vocab = {}
    try:
        with open(filepath, 'r', encoding='utf-8') as f:
            wanted_data = json.load(f)
            for tech_list in wanted_data:
                for tech in tech_list:
                    normalized = tech.lower().replace(" ", "")
                    if normalized and normalized not in standard_vocab:
                        standard_vocab[normalized] = tech
        return standard_vocab
    except Exception as e:
        print(f"원티드 단어장 로드 실패: {e}")
        return {}

standard_vocab = load_wanted_vocab()

# 2. 필수 매핑 사전 (유사도로도 못 잡는 완전 다른 표기법이나 약어들만 최소한으로 남김)
TECH_NAME_MAP = {
    'aws ecs': 'awsecs', 'aws ec2': 'awsec2', 'aws rds': 'awsrds', 'aws s3': 'awss3',
    'amazon web service': 'aws', 'google cloud platform': 'gcp',
    'postgres': 'postgresql', 'ms-sql': 'mssql', 'microsoft sql server': 'mssql', 'sql 서버': 'mssql',
    'maria db': 'mariadb', 'react-native': 'reactnative', 'github actions': 'githubactions',
    'gitlab ci': 'gitlabci', 'k8s': 'kubernetes', '쿠버네티스': 'kubernetes',
    '머신러닝': 'ml', '딥러닝': 'deeplearning', 'ci cd': 'ci/cd', 'cicd': 'ci/cd',
    'tailwind': 'tailwindcss', 'react query': 'reactquery', 'tanstack query': 'reactquery'
}

# 3. 유사도 검사 속도 최적화를 위한 캐시 딕셔너리
fuzzy_cache = {}

# 4. 데이터 클렌징 및 자동 매핑 함수
def process_combined_jobkorea_skills(row):
    def extract_list(val):
        if pd.api.types.is_scalar(val) and pd.isna(val): return []
        if isinstance(val, list): return val
        if hasattr(val, 'tolist'): return val.tolist()
        
        if isinstance(val, str):
            val = val.strip()
            if not val: return []
            if val.startswith('['):
                try: return ast.literal_eval(val)
                except:
                    items = re.findall(r"[a-zA-Z가-힣+#.-]+", val)
                    return [i for i in items if i]
            else:
                return [item.strip() for item in re.split(r'[,/·]+', val) if item.strip()]
        return []

    tech_list = extract_list(row.get('기술스택/분야', ''))
    hard_list = extract_list(row.get('hard_skills', ''))
    combined_skills = tech_list + hard_list

    cleaned_set = set()
    
    # 꼬리말 제거 정규식
    noise_pattern = re.compile(r'(개발|구축|설계|운영|관리|최적화|처리|경험|시스템|도구|아키텍처|엔지니어링|테스트|튜닝|기획|디자인|분석|과학|보안|플랫폼|서비스|모델링|자동화|인프라|역량|이해|지식|기반|환경|서빙|연동|생성|활용|수립|검증|배포|적용|리팩토링|파이프라이닝|스크래핑|작성|이관|통신|제어|파싱)$')
    
    for tech in combined_skills:
        if not tech: continue
            
        tech_str = str(tech).strip()
        cleaned_tech_str = noise_pattern.sub('', tech_str).strip()
        
        if not cleaned_tech_str: continue
            
        t_lower = cleaned_tech_str.lower()
        lookup_key = t_lower.replace(" ", "")
            
        # [Step 1] 원티드 사전에 완벽히 일치하는 경우
        if lookup_key in standard_vocab:
            cleaned_set.add(standard_vocab[lookup_key])
            continue
            
        # [Step 2] 수동 매핑 사전(TECH_NAME_MAP)에 존재하는 경우 (완전 약어/동의어)
        if t_lower in TECH_NAME_MAP:
            mapped_key = TECH_NAME_MAP[t_lower].lower().replace(" ", "")
            if mapped_key in standard_vocab:
                cleaned_set.add(standard_vocab[mapped_key])
            continue

        # [Step 3] 80% 이상 유사도 자동 매핑 (Fuzzy Matching)
        # 단어 길이가 너무 짧으면(예: C, R) 엉뚱한 매핑이 일어날 수 있으므로 4글자 이상만 허용
        if len(lookup_key) >= 4:
            # 캐시에 이미 있는 단어면 캐시 활용 (연산 속도 향상)
            if lookup_key in fuzzy_cache:
                if fuzzy_cache[lookup_key]:
                    cleaned_set.add(standard_vocab[fuzzy_cache[lookup_key]])
                continue

            # difflib.get_close_matches(단어, 비교군 리스트, 반환 갯수, 유사도 컷오프)
            # cutoff=0.8은 철자 유사도가 80% 이상일 때만 잡아내겠다는 의미
            matches = difflib.get_close_matches(lookup_key, standard_vocab.keys(), n=1, cutoff=0.8)
            
            if matches:
                best_match = matches[0]
                fuzzy_cache[lookup_key] = best_match  # 성공한 매칭 캐시에 저장
                cleaned_set.add(standard_vocab[best_match])
                # print(f"🔍 자동 복구 성공: {tech_str} -> {standard_vocab[best_match]}") # (테스트용 로깅)
            else:
                fuzzy_cache[lookup_key] = None  # 매칭 실패도 캐시에 기록하여 다음번엔 바로 버림

    return list(cleaned_set)

# --- 실행 예시 ---
df_jobkorea_location['final_techs'] = df_jobkorea_location.apply(process_combined_jobkorea_skills, axis=1)
```
