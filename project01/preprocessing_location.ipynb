```python
import pandas as pd
import re

# 테스트를 위한 임시 데이터프레임 생성
data = {
    "공고번호": ["49192253", "12345678"],
    "지역": ["", "서울 강남구"], # 하나는 비어있고, 하나는 채워져 있음
    "상세내용": [
        "...(생략)... 근무시간 주5일(월~금) 08:30 ~ 17:30 근무지주소 경남 김해시 대동면 대동산단4로 116-13 ... 경남>김해시 본 채용정보는 ...",
        "상세내용 텍스트입니다." # 이미 지역이 있으므로 처리 안 함
    ]
}
df_wanted = pd.DataFrame(data)

# 지역을 추출하는 함수
def extract_location_from_text(row):
    # 1. 이미 지역 값이 채워져 있다면 기존 값 그대로 유지
    if pd.notna(row['지역']) and str(row['지역']).strip() != '':
        return row['지역']
    
    text = str(row['상세내용'])
    
    # 2. '근무지주소 ' 뒤에 오는 '시/도 + 시/군/구' 패턴 찾기
    # 예: "근무지주소 경남 김해시..." -> "경남 김해시" 추출
    address_match = re.search(r'위치\s+([가-힣]+(?:도|광역시|특별시)?\s+[가-힣]+(?:시|군|구))', text)
    if address_match:
        return address_match.group(1)
        
    # 3. 위 패턴이 없다면 '경남>김해시' 같은 태그 패턴 찾기
    tag_match = re.search(r'([가-힣]+)>([가-힣]+(?:시|군|구))', text)
    if tag_match:
        return f"{tag_match.group(1)} {tag_match.group(2)}"
        
    # 4. 그래도 못 찾으면 NaN 반환
    return pd.NA
# apply 함수로 각 행에 적용하여 지역 컬럼 업데이트
df_jobkorea_na['지역'] = df_jobkorea_na.apply(extract_location_from_text, axis=1)

print(df_jobkorea_na[['공고번호', '지역']])
```

```python
import pandas as pd
import numpy as np
import re

# 질문에서 주신 데이터 일부 샘플링
data = [
    '울산 북구 외', '대전 서구 외', '서울 구로구 외', '서울 종로구', '서울 전지역',
    '경기 전지역 외', '세종 전지역 외', '동경도 신쥬쿠구', '효고현 고베시', 
    '충청남도 천안시', '강원특별자치도 춘천시', '<NA>', '전국', '전북 완주군'
]
df_job = pd.DataFrame({'location': data})

# 1. 시/도 명칭 통일 사전 (기존과 동일)
SIDO_MAP = {
    '서울특별시': '서울', '서울시': '서울', '서울': '서울',
    '경기도': '경기', '경기': '경기',
    '인천광역시': '인천', '인천시': '인천', '인천': '인천',
    '부산광역시': '부산', '부산시': '부산', '부산': '부산',
    '대구광역시': '대구', '대구시': '대구', '대구': '대구',
    '광주광역시': '광주', '광주시': '광주', '광주': '광주',
    '대전광역시': '대전', '대전시': '대전', '대전': '대전',
    '울산광역시': '울산', '울산시': '울산', '울산': '울산',
    '세종특별자치시': '세종', '세종시': '세종', '세종': '세종',
    '강원특별자치도': '강원', '강원도': '강원', '강원': '강원',
    '충청북도': '충북', '충북': '충북',
    '충청남도': '충남', '충남': '충남',
    '전북특별자치도': '전북', '전라북도': '전북', '전북': '전북',
    '전라남도': '전남', '전남': '전남',
    '경상북도': '경북', '경북': '경북',
    '경상남도': '경남', '경남': '경남',
    '제주특별자치도': '제주', '제주도': '제주', '제주': '제주'
}

def clean_portal_location(address):
    # 1. 완전한 결측치 및 의미 없는 광역 데이터 처리
    if pd.isna(address) or str(address).strip() in ('', '<NA>', 'None', '전국'):
        return np.nan
        
    addr_str = str(address).strip()
    
    # 2. ⭐️ 핵심: 텍스트 끝에 붙은 ' 전지역', ' 외', ' 전지역 외'를 정규식으로 한 번에 제거
    # \s*: 공백 허용, (전지역)?: 있을 수도 있고 없을 수도 있음, (외)?: 있을 수도 있고 없을 수도 있음, $: 문자열 끝
    addr_str = re.sub(r'\s*(전지역)?\s*(외)?$', '', addr_str).strip()
    
    # 3. 공백 기준으로 분리
    parts = addr_str.split()
    if not parts:
        return np.nan
        
    sido = parts[0]
    
    # 4. 시/도 표준화 및 해외 주소(동경도, 효고현 등) 필터링
    standard_sido = SIDO_MAP.get(sido)
    
    # 사전에 없는 지역(해외 또는 오타)은 가차없이 버림 (지도 시각화 시 오류 방지)
    if not standard_sido:
        return np.nan
        
    # 5. 최종 텍스트 조립
    if len(parts) >= 2:
        sigungu = parts[1].replace(',', '')
        return f"{standard_sido} {sigungu}"
    else:
        # '서울 전지역'이 '서울'로 잘려나간 경우 여기로 들어옴
        return standard_sido

# 전처리 적용
df_jobkorea_na['location_clean'] = df_jobkorea_na['지역'].apply(clean_portal_location)

print("=== 잡코리아/사람인 패턴 전처리 결과 ===")
print(df_jobkorea_na[['지역', 'location_clean']])
```

```python
import pandas as pd

# 테스트용 예시 데이터셋


# 시/도 명칭을 2글자로 통일하기 위한 매핑 사전
SIDO_MAP = {
    '서울특별시': '서울', '서울시': '서울', '서울': '서울',
    '경기도': '경기', '경기': '경기',
    '인천광역시': '인천', '인천시': '인천', '인천': '인천',
    '부산광역시': '부산', '부산시': '부산', '부산': '부산',
    '대구광역시': '대구', '대구시': '대구', '대구': '대구',
    '광주광역시': '광주', '광주시': '광주', '광주': '광주',
    '대전광역시': '대전', '대전시': '대전', '대전': '대전',
    '울산광역시': '울산', '울산시': '울산', '울산': '울산',
    '세종특별자치시': '세종', '세종시': '세종', '세종': '세종',
    '강원특별자치도': '강원', '강원도': '강원', '강원': '강원',
    '충청북도': '충북', '충북': '충북',
    '충청남도': '충남', '충남': '충남',
    '전북특별자치도': '전북', '전라북도': '전북', '전북': '전북',
    '전라남도': '전남', '전남': '전남',
    '경상북도': '경북', '경북': '경북',
    '경상남도': '경남', '경남': '경남',
    '제주특별자치도': '제주', '제주도': '제주', '제주': '제주'
}

# 자주 나오는 구 이름을 미리 매핑해두기 (성남시 구별 상세 추가)
GU_TO_SIDO = {
    '강남구': '서울 강남구', '서초구': '서울 서초구', '마포구': '서울 마포구', 
    '영등포구': '서울 영등포구', '구로구': '서울 구로구', '금천구': '서울 금천구',
    '분당구': '경기 성남시 분당구', '수정구': '경기 성남시 수정구', '중원구': '경기 성남시 중원구', '판교': '경기 성남시 분당구'
}

def clean_wanted_location(address):
    # 1. 결측치 처리
    if pd.isna(address) or str(address).strip() == '':
        return None
        
    # 2. 공백을 기준으로 텍스트 분리
    parts = str(address).strip().split()
    
    # 3. 주소가 2단어 이상인 경우 (예: "서울특별시 서초구 ...")
    if len(parts) >= 2:
        sido = parts[0]
        
        # 접두사가 구 이름으로 시작하는 예외 처리 (예: "분당구 판교역로")
        if sido in GU_TO_SIDO:
            return GU_TO_SIDO[sido]
            
        sigungu = parts[1].replace(',', '') # 시군구 뒤에 혹시 붙어있을 쉼표 제거
        
        # 시/도 이름 통일 (사전에 없으면 원래 이름 유지)
        sido_standard = SIDO_MAP.get(sido, sido)
        
        # ⭐️ 성남시 예외 처리: 시군구가 '성남시'이고, 뒤에 단어가 더 있다면 구 이름까지 추출
        if sigungu == '성남시' and len(parts) >= 3:
            gu = parts[2].replace(',', '')
            return f"{sido_standard} {sigungu} {gu}"
        
        return f"{sido_standard} {sigungu}"
        
    # 4. 주소가 1단어밖에 없는 예외 상황
    elif len(parts) == 1:
        sido = parts[0]
        if sido in GU_TO_SIDO:
            return GU_TO_SIDO[sido]
        return SIDO_MAP.get(sido, sido)
        
    return address

# apply 함수를 사용해 전처리 적용 및 새로운 컬럼 생성
df_wanted_na['location_clean'] = df_wanted_na['location'].apply(clean_wanted_location)

print("=== 전처리 결과 ===")
print(df_wanted_na[['location', 'location_clean']])
```

```python
import pandas as pd

# 1. 대한민국 시/도 명칭으로 시작할 수 있는 모든 경우의 수(접두사)를 튜플로 정의합니다.
valid_sido_prefixes = (
    '서울', '경기', '인천', '부산', '대구', '광주', '대전', '울산', '세종', 
    '강원', '충북', '충청북', '충남', '충청남', '전북', '전라북', '전남', '전라남', 
    '경북', '경상북', '경남', '경상남', '제주'
)

# 2. 결측치를 제외하고(na=False), 위 접두사로 시작하지 않는(~) 행만 필터링합니다.
# 주의: 이 필터링은 원본 데이터인 'location' 컬럼을 기준으로 확인하는 것이 좋습니다.
outlier_mask = ~df_wanted_na['location_clean'].str.startswith(valid_sido_prefixes, na=False)

# 3. 추가로 빈칸이나 완전한 결측치(NaN)가 아닌 경우만 뽑아내어 진짜 '이상한 주소'만 남깁니다.
not_empty_mask = df_wanted_na['location_clean'].astype(str).str.strip() != ''
not_na_mask = df_wanted_na['location_clean'].notna()

final_outlier_mask = outlier_mask & not_empty_mask & not_na_mask

# 예외 데이터 확인
df_outliers = df_wanted_na[final_outlier_mask]

print(f"찾아낸 예외 데이터 개수: {len(df_outliers)}개")
print("=== 예외 데이터 목록 ===")
print(df_outliers[['location_clean']])
df_outliers['location_clean'].unique()
```

```python
LAST_CHANCE_MAP = {
    '금토로 80번길': '경기 성남시 수정구',
    '성남시 분당구': '경기 성남시 분당구',
    '성남시 창업로': '경기 성남시 수정구',
    '분당스퀘어(서현역)': '경기 성남시 분당구',
    '서울 강서구마곡중앙6로': '서울 강서구',
    '서울 강남대로84길': '서울 강남구',
    '서울 강남대로53길': '서울 서초구', # 강남대로는 서초/강남이 섞여있지만 임의 지정
    '동구 중앙대로': '부산 동구', # 문맥상 부산일 확률이 높음 (또는 대구)
    '경기 성남시 판교역로184': '경기 성남시 분당구'
}

def final_polish(address):
    # 1. 결측치 또는 'None' 문자열 날리기 (NaN 반환)
    if pd.isna(address) or str(address) == 'None' or str(address).strip() == '':
        return np.nan
        
    addr_str = str(address).strip()
    
    # 2. 마지막 매핑 사전에 있으면 바로 치환
    for key, value in LAST_CHANCE_MAP.items():
        if key in addr_str:
            return value
            
    # 3. '원격근무'는 살려둠
    if addr_str == '원격근무':
        return addr_str
        
    # 4. 정규식을 이용해 텍스트 다듬기
    # 예: '서울 마포대로' -> '서울 마포구' 로 치환하기 어려운 경우, 
    # '마포'라는 단어가 들어가면 마포구로 강제 변환하는 식의 로직 추가
    if '서울' in addr_str and '마포' in addr_str: return '서울 마포구'
    if '서울' in addr_str and '강남' in addr_str: return '서울 강남구'
    if '서울숲' in addr_str or '성수' in addr_str: return '서울 성동구'
    
    # 5. 형식 검사: "시/도 + 띄어쓰기 + 구/군/시" 형태가 아니면 버림 (NaN 처리)
    # 정규식 패턴: '서울', '경기' 등 2글자 단어로 시작하고 띄어쓰기 후 '구', '시', '군'으로 끝나는 패턴
    # 예: ^(서울|경기|인천|부산|대구|광주|대전|울산|세종|강원|충북|충남|전북|전남|경북|경남|제주)\s+[가-힣]+(구|시|군)$
    # 단, '경기 성남시 분당구' 처럼 3단어인 경우도 허용하도록 유연하게 작성
    import re
    valid_pattern = r'^(서울|경기|인천|부산|대구|광주|대전|울산|세종|강원|충북|충남|전북|전남|경북|경남|제주)\s+[가-힣\s]+(구|시|군)$'
    
    if re.match(valid_pattern, addr_str):
        return addr_str # 완벽한 형태면 통과
    elif addr_str in ('서울', '경기', '부산', '대구', '인천', '광주', '대전', '울산', '제주'):
        # '서울', '경기' 처럼 광역단위만 있는 것도 살리고 싶다면 통과 (버리고 싶으면 np.nan)
        return addr_str
    else:
        # '3F, 343', '서울 장충단로' 등 패턴에 안 맞는 찌꺼기는 가차없이 버림
        return np.nan

# 새로운 컬럼에 최종 정제 결과 저장
df_wanted_na['location_perfect'] = df_wanted_na['location_final'].apply(final_polish)

# 의미 없는 데이터(NaN)를 아예 데이터프레임에서 삭제(Drop)
# 원격근무나 특정 지역 공고만 남습니다.
df_wanted_clean = df_wanted_na.dropna(subset=['location_perfect'])

print("=== 살아남은 완벽한 데이터 목록 ===")
print(df_wanted_clean['location_perfect'].unique())

print(f"\n버려진 데이터 수: {len(df_wanted_na) - len(df_wanted_clean)}개")
```
