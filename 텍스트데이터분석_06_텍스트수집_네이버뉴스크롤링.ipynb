# 네이버 검색 API 예제 코드

```python
import os
import urllib.request
from dotenv import load_dotenv
load_dotenv()
client_id = os.getenv("CLIENT_ID")
client_secret = os.getenv("CLIENT_SECRET")

encText = urllib.parse.quote("인공지능") # 검색어를 URL 인코딩하여 안전하게 전달할 수 있도록 함

start = 1
display = 10
while start <= 20: # 10개씩 반복해서 가져올 수 있게하는 반복문

    url = "https://openapi.naver.com/v1/search/blog?query=" + encText # JSON 결과
    url += f'&start={start}&display={display}' # 옵션 추가
    # url = "https://openapi.naver.com/v1/search/blog.xml?query=" + encText # XML 결과

    request = urllib.request.Request(url) # HTTP 요청 객체 생성
    request.add_header("X-Naver-Client-Id",client_id) 
    request.add_header("X-Naver-Client-Secret",client_secret)
    response = urllib.request.urlopen(request) # HTTP 요청 보내고 응답 받기
    rescode = response.getcode() # HTTP 응답 코드 확인
    if(rescode==200): # HTTP 응답 코드가 200이면 성공적으로 데이터를 가져온 것
        news_data = response.read().decode('utf-8')
        print(news_data)
    else:
        print("Error Code:" + rescode)
        break
    start += display
```

```python
import json
import pandas as pd
py_news_data = json.loads(news_data) # JSON 문자열을 Python 객체로 변환
json_string = json.dumps(py_news_data, indent=4, ensure_ascii=False) # Python 객체를 JSON 문자열로 변환 한글일때는 False, indent는 들여쓰기 설정
print(json_string) # JSON 문자열 출력
# pyobject = json.loads(json_string) # JSON 문자열을 Python 객체로 변환
# json.dump(pyobject, open('news_data.json', 'w'), indent=4, ensure_ascii=False) # Python 객체를 JSON 파일로 저장
type(py_news_data)
```

```python
def crawlNaverNews(url, start=1, display=10):
    url += f'&start={start}&display={display}'

    request = urllib.request.Request(url)
    request.add_header("X-Naver-Client-Id",client_id) 
    request.add_header("X-Naver-Client-Secret",client_secret)
    
    response = urllib.request.urlopen(request)
    rescode = response.getcode() 

    if(rescode==200): # HTTP 응답 코드가 200이면 성공적으로 데이터를 가져온 것
        news_data = response.read().decode('utf-8')
        return json.loads(news_data) # JSON 문자열을 Python 객체로 변환하여 반환
    else:
        print("Error Code:" + rescode)
        return None
```

```python
def mergeResultToList(result_list, merged_list):
    if result_list is not None:
        merged_list += result_list['items'] # 가져온 뉴스 데이터를 뉴스 코퍼스에 추가
        return True
    else: return False
```

```python
def saveJsonToCSV(merged_list, csv_filename):
    df = pd.DataFrame(merged_list) # 리스트를 DataFrame으로 변환
    df.to_csv(f'result/{csv_filename}', index=False, encoding='utf-8-sig')
```

```python
def crawlMain():
    encText = urllib.parse.quote('인공지능')
    url = "https://openapi.naver.com/v1/search/news?query=" + encText
    news_corpus = []
    start = 1
    display = 10

    while start <= 100:
        news_data = crawlNaverNews(url, start, display)
        mer = mergeResultToList(news_data, news_corpus)
        if not mer: break
        start += display
    print(news_corpus)
    saveJsonToCSV(news_corpus, 'news_corpus.csv')

crawlMain()
```
