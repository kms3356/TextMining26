## 텍스트 특징 추출 (TF-IDF)

```python
sample_corpus = [
    '자연어처리 강의를 시작하겠습니다.', 
    '자연어처리는 재미있습니다.', 
    '밥을 먹고 강의를 듣고 있습니다.', 
    '이번 자연어처리 강의는 한국어 자연어처리입니다.' 
    ]
```

```python
import sklearn
```

```python
from konlpy.tag import Okt
def my_tokenizer(text):

    mytags = set(['Noun', 'Verb', 'Adjective'])
    # 불용어 삭제 : 의미없는데 고빈도로 나오는 단어 추가
    my_stopwords = set('하는 한다 의하여 하여 있다 하며 하여야'.split())
    res = [word for word, tag in Okt().pos(text) if tag in mytags and word not in my_stopwords and len(word) > 1]
    return res
```

```python
# TFidfVectorizer 객체 색성
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer(tokenizer=my_tokenizer)
# 특징 집합과 관련 데이터 모델을 생성
vectorizer.fit(sample_corpus) # 학습
print(vectorizer.get_feature_names_out())
# 특징 벡터 추출
sample_dtm = vectorizer.transform(sample_corpus) # 문제풀기
sample_dtm.toarray()
```

### 다음영화 실습

```python
import pandas as pd
datafile = 'data/daum_movie_review.csv'
df = pd.read_csv(datafile)
vectorizer = TfidfVectorizer(tokenizer=my_tokenizer, max_features=100)
# 특징 집합과 관련 데이터 모델을 생성
vectorizer.fit(df['review'])
print(vectorizer.get_feature_names_out())
# 특징 벡터 추출
sample_dtm = vectorizer.transform(df['review'])
print(sample_dtm.toarray())
```
